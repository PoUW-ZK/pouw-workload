# Arrival Models

Compare uniform, Poisson, burst, and wave-based request arrival models over a 100-second window.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
```

```python
def simulate_uniform(duration_seconds, base_rate):
    """Return a constant request count for every one-second interval."""
    times = np.arange(duration_seconds)
    counts = np.full(duration_seconds, base_rate, dtype=float)
    return times, counts


def simulate_poisson(duration_seconds, base_rate):
    """Return independent Poisson request counts for each second."""
    times = np.arange(duration_seconds)
    counts = np.random.poisson(base_rate, size=duration_seconds)
    return times, counts


def simulate_bursts(duration_seconds, base_rate, burst_period=20):
    """Alternate between high-rate and low-rate periods."""
    times = np.arange(duration_seconds)
    high_rate = base_rate * 2
    low_rate = base_rate * 0.25
    in_burst = (times % burst_period) < (burst_period // 2)
    rates = np.where(in_burst, high_rate, low_rate)
    counts = np.random.poisson(rates)
    return times, counts


def simulate_waves(duration_seconds, base_rate, period=50):
    """Modulate the base rate with a sine wave."""
    times = np.arange(duration_seconds)
    rates = base_rate * (1 + 0.75 * np.sin(2 * np.pi * times / period))
    counts = np.random.poisson(np.maximum(rates, 0))
    return times, counts
```

```python
duration_seconds = 100
base_rate = 10

models = [
    ("Uniform", simulate_uniform),
    ("Poisson", simulate_poisson),
    ("Bursts", simulate_bursts),
    ("Waves", simulate_waves),
]

fig, axes = plt.subplots(2, 2, figsize=(14, 9), sharex=True)
for axis, (name, model) in zip(axes.flat, models):
    times, counts = model(duration_seconds, base_rate)
    axis.step(times, counts, where="mid")
    axis.set_title(f"{name} arrival")
    axis.set_xlabel("Time (seconds)")
    axis.set_ylabel("Requests per second")
    axis.grid(alpha=0.3)

fig.suptitle("Request Arrival Models (base rate: 10 requests/second)")
fig.tight_layout()
plt.show()
```
